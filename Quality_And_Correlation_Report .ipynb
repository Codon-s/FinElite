# Outlier Analysis

Using the **Interquartile Range (IQR)** method:

| Feature                      | Outliers Detected |
| ---------------------------- | ----------------: |
| Avg_Monthly_Spending         |            **23** |
| Number_of_Defaults           |           **319** |
| Credit_Limit                 |             **1** |
| Remaining Numerical Features |             **0** |

### Interpretation

**Number_of_Defaults**

* Highest number of detected outliers.
* Likely represents customers with unusually high default counts.
* These records may be genuine high-risk customers rather than errors.

**Avg_Monthly_Spending**

* Contains a small number of extreme spending values.
* Could represent high-spending customers or exceptional cases.

**Credit_Limit**

* Only one outlier detected.
* Minimal impact on overall analysis.

### Recommendation

Instead of removing all outliers:

* Investigate extreme values.
* Use winsorization or robust scaling where appropriate.
* Retain genuine business outliers if they represent important customer segments.
