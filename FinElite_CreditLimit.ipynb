```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

**Read the Excel Dataset**

```python
df = pd.read_excel('../Datasets/Credir_Card_Bank.xlsx')
pd.set_option("display.max_columns",None)
df
```

**Dataset Overview, Data Types & Missing Value Analysis**

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

### **Data Preprocessing**
**Correlation Analysis**

```python
df.corr(numeric_only=True)
```

**Remove Irrelevant Features & Encode Categorical Variables**

```python
df = df.drop(columns=["Customer_ID","Monthly_Income","PAN_Verified", "KYC_Status"])
df
```

```python
df = pd.get_dummies(df,columns=['Gender','Employment_Type','Occupation','Residential_Status','Fraud_Flag'],drop_first=True)
print(df.shape)
df.head()
```

**Separated Features and Target Variable**

```python
x = df.drop(['Credit_Limit'],axis=1)
y = df['Credit_Limit']

print("Features shape:", x.shape)
print("Target shape:", y.shape)

print("\nTarget Distribution: Credit Limit")
print(y)
```

**Split Data into Training and Testing Sets**

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2, random_state=42)
```

## **ML Model Fitting & Regression Analysis**
- **Linear Regression**
- **Random Forest Regression**
- **XGBoost Regression**

```python
from sklearn.linear_model import LinearRegression
modellinear= LinearRegression()
modellinear.fit(xtrain,ytrain)
yplinear= modellinear.predict(xtest)
modellinear
```

```python
from sklearn.ensemble import RandomForestRegressor
modelR=RandomForestRegressor(random_state=42)
modelR.fit(xtrain,ytrain)
ypR=modelR.predict(xtest)
modelR
```

```python
from xgboost import XGBRegressor
modelxgbr = XGBRegressor(random_state=42) 
modelxgbr.fit(xtrain, ytrain) 
ypxgbr = modelxgbr.predict(xtest)
modelxgbr
```

# **Metrics Evaluation & 5-Fold Cross-Validation**

### **Linear Regression**

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
Linear_R2score = r2_score(ytest , yplinear)
lrMAE = mean_absolute_error(ytest, yplinear)
lrMSE = mean_squared_error(ytest, yplinear)
lrRMSE = np.sqrt(mean_squared_error(ytest, yplinear))

print("Linear_R2score :", Linear_R2score*100)
print("Mean Absolute Error:", lrMAE)
print("Mean Squared Error:", lrMSE)
print("Root_Mean Squared Error:", lrRMSE)
```

```python
from sklearn.model_selection import cross_val_score
cvlinear = LinearRegression()
linear_cvscore = cross_val_score(cvlinear,x,y,cv=5,scoring='r2')*100
print(linear_cvscore)
print('linear_cvscore',linear_cvscore.mean())
print("Standard Deviation:",linear_cvscore.std())
```

### **Random Forest Regression**

```python
Random_R2score = r2_score(ytest , ypR)
rfMAE = mean_absolute_error(ytest, ypR)
rfMSE = mean_squared_error(ytest, ypR)
rfRMSE = np.sqrt(mean_squared_error(ytest, ypR))

print("Random_R2score:", Random_R2score*100)
print("Mean Absolute Error:", rfMAE)
print("Mean Squared Error:", rfMSE)
print("Root_Mean Squared Error:", rfRMSE)
```

```python
cvRandom = RandomForestRegressor()
Random_cvscore = cross_val_score(cvRandom,x,y,cv=5,scoring='r2')*100
print(Random_cvscore)
print('Random_cvscore',Random_cvscore.mean())
```

### **XGBoost Regression**

```python
XGB_R2score = r2_score(ytest , ypxgbr)
xgMAE = mean_absolute_error(ytest, ypxgbr)
xgMSE = mean_squared_error(ytest, ypxgbr)
xgRMSE = np.sqrt(mean_squared_error(ytest,ypxgbr))

print("XGBoost_R2score:", XGB_R2score*100)
print("Mean Absolute Error:", xgMAE)
print("Mean Squared Error:", xgMSE)
print("Root_Mean Squared Error:", xgRMSE)
```

```python
cvXGB= XGBRegressor()
XGB_cvscore = cross_val_score(cvXGB,x,y,cv=5,scoring='r2')*100
print(XGB_cvscore)
print('XGBoost_cvscore',XGB_cvscore.mean())
```

## **Hyperparameter Tuning using GridSearchCV**
- **Random Forest & XGBoost Regression**

```python
from sklearn.model_selection import GridSearchCV
param_grid_rf = {
    'n_estimators': [100, 200],
    'max_depth': [None, 5, 10],
    'min_samples_split': [2, 5]
}

grid_rf = GridSearchCV(
    estimator=modelR,
    param_grid=param_grid_rf,
    cv=5,
    scoring='r2',
    n_jobs=-1)

grid_rf.fit(xtrain,ytrain)
grid_rf.best_estimator_

print(" Random Forest Regression \n")
print("Best Parameters:")
print(grid_rf.best_params_)
print("Best R2 Score:", grid_rf.best_score_*100)
```

```python
tunned_params_xgb = {'n_estimators': [100, 200],
                         'learning_rate': [0.05, 0.1],
                         'max_depth': [3, 5],
                         'subsample': [0.8, 1.0],
                         'colsample_bytree': [0.8, 1.0]}

params_xgb = GridSearchCV(
    estimator = modelxgbr,param_grid = tunned_params_xgb,scoring='r2',cv=5,n_jobs=-1
)

params_xgb.fit(xtrain, ytrain)
params_xgb.best_estimator_

print("XGBoost Regression \n")
print("Best Parameters:")
print(params_xgb.best_params_)
print("Best R2 Score:", params_xgb.best_score_*100)
```

### **Model Performance Comparison** 
- **Best Model Selection & Hyperparameter**

```python
results = pd.DataFrame({
    "Model": ['Linear Regression','Random Forest Regressor','XGBoost Regressor'],
    "R2 Score": [Linear_R2score*100,Random_R2score*100,XGB_R2score*100],
    "Cross_Validation_Score": [linear_cvscore.mean(),Random_cvscore.mean(),XGB_cvscore.mean()],
    "MAE": [lrMAE,rfMAE,xgMAE],
    "MSE": [lrMSE,rfMSE,xgMSE ],
    "RMSE": [lrRMSE,rfRMSE,xgRMSE],
})

results = results.sort_values(by=["R2 Score", "Cross_Validation_Score","MAE", "MSE", "RMSE"], ascending=[False, False, True, True, True]).reset_index(drop=True)
results
```

```python
results2 = results.sort_values(
    by='R2 Score',
    ascending=False
).reset_index(drop=True)

results2["Best Parameters"] = results2["Model"].map({
    "Linear Regression": "Default Parameters",
    "Random Forest Regressor": str(grid_rf.best_params_),
    "XGBoost Regressor": str(params_xgb.best_params_)
})

best_model_name = results2.iloc[0]["Model"]

print("Best Model:", best_model_name)

results2
```

- **Linear Regression achieved near-perfect performance (R² ≈ 99.99–100%) with consistent cross-validation results Annual Income (~0.80 correlation) is the strongest predictor of Credit Limit in the dataset**
- **XGBoost (98.64%) and Random Forest (97.31%) also delivered strong predictive performance**
- **However, their higher MAE, MSE, and RMSE indicate greater prediction errors than Linear Regression,
